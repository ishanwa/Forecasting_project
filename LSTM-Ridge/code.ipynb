```python
!pip install tensorflow openpyxl
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob, os
```

```python
from sklearn.preprocessing import MaxAbsScaler
from sklearn.metrics import mean_absolute_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.regularizers import l2
```

```python
data_folder   = r"/content"
output_folder = r"/content"

os.makedirs(output_folder, exist_ok=True)

files = glob.glob(os.path.join(data_folder, "RevenueCenter_*.xlsx"))
```

```python
# Error matrix
def mape(y_true, y_pred):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    return np.mean(np.abs((y_true - y_pred) / (y_true + 1e-8))) * 100
```

```python

def create_sequences(series, window=30):
    X, y = [], []
    for i in range(len(series) - window):
        X.append(series[i:i+window])
        y.append(series[i+window])
    return np.array(X), np.array(y)
```

```python
for file in files:

    center_name = os.path.basename(file).replace(".xlsx", "")
    print("\nProcessing:", center_name)

    center_path = os.path.join(output_folder, center_name)
    os.makedirs(center_path, exist_ok=True)

    df = pd.read_excel(file)
    df["Date"] = pd.to_datetime(df["Date"])
    df = df.sort_values("Date").reset_index(drop=True)

    # LAST 2 MONTHS AS TEST
    split_date = df["Date"].max() - pd.DateOffset(months=2)

    meals = ["BreakFast", "Lunch", "Dinner"]

    for meal in meals:

        temp_df = df[["Date", meal]].copy()
        temp_df.rename(columns={meal: "y"}, inplace=True)

        train_df = temp_df[temp_df["Date"] <= split_date]
        test_df  = temp_df[temp_df["Date"] > split_date]

        if train_df["y"].nunique() <= 1:
            print(f" Skipping {center_name} - {meal}: target is constant")
            continue

        # SCALING
        scaler = MaxAbsScaler()
        train_scaled = scaler.fit_transform(train_df[["y"]])
        test_scaled  = scaler.transform(test_df[["y"]])

        # SEQUENCES (FIXED)
        WINDOW = 30

        # Training sequences
        X_train, y_train = create_sequences(train_scaled, WINDOW)

        # Combine last WINDOW train + full test
        combined = np.vstack([train_scaled[-WINDOW:], test_scaled])
        X_test, y_test = create_sequences(combined, WINDOW)

        X_train = X_train.reshape((X_train.shape[0], WINDOW, 1))
        X_test  = X_test.reshape((X_test.shape[0],  WINDOW, 1))

        # MODEL WITH RIDGE REGULARIZATION (L2)
        model = Sequential([
            LSTM(64,
                 input_shape=(WINDOW, 1),
                 kernel_regularizer=l2(0.01),  # Ridge regularization on weights
                 recurrent_regularizer=l2(0.01),  # Ridge regularization on recurrent weights
                 bias_regularizer=l2(0.01)),  # Ridge regularization on biases
            Dense(1, kernel_regularizer=l2(0.01))  # Ridge on output layer
        ])

        model.compile(
            optimizer="adam",
            loss="mse"
        )

        model.fit(
            X_train, y_train,
            epochs=100,
            batch_size=32,
            verbose=0,
            validation_split=0.2  # Added for better regularization monitoring
        )

        # PREDICTION
        y_train_pred = scaler.inverse_transform(model.predict(X_train))
        y_test_pred  = scaler.inverse_transform(model.predict(X_test))

        y_train_true = scaler.inverse_transform(y_train.reshape(-1,1))
        y_test_true  = scaler.inverse_transform(y_test.reshape(-1,1))

        # METRICS
        train_mae  = mean_absolute_error(y_train_true, y_train_pred)
        train_r2   = r2_score(y_train_true, y_train_pred)
        train_mape = mape(y_train_true, y_train_pred)

        test_mae  = mean_absolute_error(y_test_true, y_test_pred)
        test_r2   = r2_score(y_test_true, y_test_pred)
        test_mape = mape(y_test_true, y_test_pred)

        # DATES (FULL 2 MONTHS)
        test_dates = test_df["Date"]

        # PLOT
        plt.figure(figsize=(10, 5))

        plt.plot(test_dates, y_test_true, label="Actual", linewidth=2)
        plt.plot(test_dates, y_test_pred, "--", label="Predicted", linewidth=2)

        plt.title(
            f"{center_name} - {meal} (LSTM with Ridge Regularization)\n"
            f"TRAIN → MAE: {train_mae:,.0f} | R²: {train_r2:.3f} | MAPE: {train_mape:.1f}%\n"
            f"TEST  → MAE: {test_mae:,.0f} | R²: {test_r2:.3f} | MAPE: {test_mape:.1f}%",
            fontsize=11
        )

        plt.xlabel("Date")
        plt.ylabel("Revenue")
        plt.legend()
        plt.grid(alpha=0.3)
        plt.xticks(rotation=45)
        plt.tight_layout()

        """
        plot_file = os.path.join(center_path, f"{meal}_lstm_ridge.png")
        plt.savefig(plot_file, dpi=300)
        """
        plt.show()
        plt.close()

    print(f"Saved LSTM-Ridge results → {center_path}")

print("\nALL REVENUE CENTERS COMPLETED (LSTM with Ridge Regularization)")
```
